## 1 · El programa completo

Tres ingredientes: una tabla de patrones, un mapa de sustitución de pronombres
y una lista de respuestas por defecto. Eso es todo lo que ELIZA fue siempre.


```python
import re
import random

random.seed(1)   # reproducible en clase; elimínalo para improvisar en vivo

RULES = [
    (r"\bI am (.*)", ["How long have you been {0}?",
                      "Why do you say you are {0}?"]),
    (r"\bI feel (.*)", ["Do you often feel {0}?"]),
    (r"\bmy (mother|father|family)\b", ["Tell me more about your {0}."]),
    (r"\bbecause\b", ["Is that the real reason?"]),
    (r"\bno\b", ["Why not?"]),
    (r"\b(computer|machine)s?\b", ["Do machines worry you?"]),
]

FALLBACK = ["Please go on.", "What does that suggest to you?",
            "Can you elaborate on that?"]

REFLECT = {"i": "you", "me": "you", "my": "your", "am": "are",
           "you": "I", "your": "my", "are": "am"}

# Busca la palabra en REFLECT, SEPARRA EL TEXTO EN LISTA
def reflect(text):
    return " ".join(REFLECT.get(w, w) for w in text.lower().split())


def respond(text):
    for pattern, templates in RULES:
        # La regla que quiero buscar, el texto donde busco, coincidencia a pesar de mayusculas y min
        m = re.search(pattern, text, re.IGNORECASE)
        if m:
            #Parte de lo que sigue despues de la coincidencia m.groups()
            groups = [reflect(g) for g in m.groups()]
            # *groups rellena los 0 con lo que sigue despues de la coincidencia
            return random.choice(templates).format(*groups)
    return random.choice(FALLBACK)


print("ELIZA tiene", sum(len(t) for _, t in RULES) + len(FALLBACK), "respuestas predefinidas.")
```

## 2 · Funciona

Ejecuta esta celda. Pide que te envíen entradas por el chat. Durante unos seis intercambios,
las respuestas parecerán inquietantemente plausibles: justo el tiempo que necesitaron
los usuarios de Weizenbaum para desarrollar vínculos afectivos.


```python
conversation = [
    "I am not feeling great about the exam",
    "I feel anxious most of the time",
    "My mother thinks I should study more",
    "I did badly because I did not sleep",
]

for line in conversation:
    print(f"tú   > {line}")
    print(f"ELIZA> {respond(line)}\n")
```

## 3 · Ahora hazla fallar: ese es el objetivo del ejercicio

Cuatro fallos, cada uno de un *tipo* distinto. Lee las anotaciones antes de ejecutar
la celda: predecir el tipo de fallo es más valioso que observarlo.


```python
probes = [
    ("I am not feeling great about the exam",
     "FUNCIONA — se detectó el patrón y se sustituyeron los pronombres"),
    ("Do you think I should drop this course?",
     "SIN MODELO DE LA PREGUNTA — una petición directa de consejo recibe una evasiva"),
    ("My mother is a doctor and my father is a lawyer",
     "IGNORA LA MITAD DE LA ORACIÓN — gana la primera coincidencia; el resto es invisible"),
    ("I am my own worst enemy",
     "PROFUNDIDAD ACCIDENTAL — sintaxis perfecta, significado vacío"),
]

random.seed(1)
for text, note in probes:
    print(f"tú   > {text}")
    print(f"ELIZA> {respond(text)}")
    print(f"        ^ {note}\n")
```

## 4 · Pruébalo tú mismo
